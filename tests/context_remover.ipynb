```python
# Imports
import os, json
import matplotlib.pyplot as plt
from sqlalchemy import create_engine
from sqlalchemy.orm import sessionmaker
import numpy as np
import seaborn as sns
```

```python
notebook_path = os.getcwd()
project_root = os.path.dirname(os.path.abspath(notebook_path))
data_dir = os.path.join(project_root, "data")
evaluations_dir = os.path.join(data_dir, "evaluations")
generated_dir = os.path.join(data_dir, "generated")
vocab_dir = os.path.join(data_dir, "vocab")
other_dir = os.path.join(data_dir, "other")
db_path = os.path.join(data_dir, "DB.db")
```

```python
engine = create_engine(f"sqlite:///{db_path}")
session_factory = sessionmaker(bind=engine)
session = session_factory()
```

```python
unique_ids= [
    "2bedd5bb-4bd4-455d-ba18-d0bb214037fc", #1 gemma 2:9b
    "b713aeeb-c613-4d58-a2ac-4a62997db744", #2 llama 3:8b
    "e1291b3b-e452-4b64-8170-eec3ed0b120c", #3 llama 3.1:8b
    "f17a95c0-47e5-4beb-a074-17db3b9755a7", #4 mistral nemo:12b
]
```

```python
res = []
for unique_id in unique_ids:
    file_path = os.path.join(generated_dir, f"{unique_id}.json")
    with open(file_path, "r", encoding="utf-8") as file:
        res.append(json.load(file))
```

```python
for i in range(len(res)):
    for j in range(len(res[i]['responses'])):
            del res[i]['responses'][j]['raw']['context']
```

```python
for output in res:
    uid = output['unique_id']
    path = os.path.join(other_dir, f"{uid}.json")
    with open(path, "w", encoding="utf-8") as file:
        json.dump(output, file, indent=4, ensure_ascii=False)
```
