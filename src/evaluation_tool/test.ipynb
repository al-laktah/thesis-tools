```python
# Imports
import os
from sqlalchemy import create_engine
from sqlalchemy.orm import sessionmaker
from db import CorrectReports
from util import save_to_json
import pandas as pd
```

```python
# Paths
notebook_path = os.getcwd()
project_root = os.path.dirname(os.path.dirname(os.path.abspath(notebook_path)))
data_dir = os.path.join(project_root, "data")
db_path = os.path.join(data_dir, "DB.db")
```

```python
# SQL Alchemy session
engine = create_engine(f"sqlite:///{db_path}")
session_factory = sessionmaker(bind=engine)
session = session_factory()
```

```python
gtb = {
    "model": {
        "id": 0,
        "name": "GT-Befund",
        "size": "NA",
        "options": None,
        "family": "gt",
        "short_name": "gtb"
    },
    "prompt": {
        "id": 0,
        "text": None,
        "tags": None
    },
    "unique_id": "gtb.45af4a46-d258-4251-b734-d407602299bd",
    "responses": [],
}
```

```python
df = CorrectReports.get_befunds(session)
```

```python
print(df.head())
```

```python
i = 1
for index, row in df.iterrows():
    if i == row["ris_id"]:
        gtb["responses"].append({
            "ris_id": row["ris_id"],
            "raw": {
                "response": row["befund"],
            }
        })
        #print(i, row["ris_id"])
        i += 1
    else:
        while i != row["ris_id"]:
            gtb["responses"].append({
                "ris_id": i,
                "raw": {
                    "response": None,
                }
            })
            i += 1
        gtb["responses"].append({
            "ris_id": row["ris_id"],
            "raw": {
                "response": row["befund"],
            }
        })
        #print(i, row["ris_id"])
        i += 1
```

```python
save_to_json(gtb, "gtb.45af4a46-d258-4251-b734-d407602299bd.json")
```
